```python
import pandas as pd

train_df = pd.read_csv("train_df.csv")
test_df = pd.read_csv("test_df.csv")
sample_submission = pd.read_csv("sample_submission.csv")

print("Train Shape:", train_df.shape)
print("Test Shape:", test_df.shape)
print("Sample Submission Shape:", sample_submission.shape)
```

```python
print("TRAIN DATA")
print(train_df.columns.tolist())

print("\nTEST DATA")
print(test_df.columns.tolist())

print("\nSAMPLE SUBMISSION")
print(sample_submission.columns.tolist())
```

```python
print("Data Types:")
print(train_df.dtypes)

print("\nMissing Values:")
print(train_df.isnull().sum())

print("\nReadmission Count:")
print(train_df["readmitted"].value_counts())
```

```python
print("Gender:")
print(train_df["gender"].unique())

print("\nPrimary Diagnosis:")
print(train_df["primary_diagnosis"].unique())

print("\nDischarge To:")
print(train_df["discharge_to"].unique())
```

```python
X = train_df.drop("readmitted", axis=1)
y = train_df["readmitted"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical columns
numeric_features = [
    "age",
    "num_procedures",
    "days_in_hospital",
    "comorbidity_score"
]

# Categorical columns
categorical_features = [
    "gender",
    "primary_diagnosis",
    "discharge_to"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Apply preprocessing
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Original X_train shape:", X_train.shape)
print("Processed X_train shape:", X_train_processed.shape)

print("Original X_test shape:", X_test.shape)
print("Processed X_test shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    penalty="l2",
    C=1.0,
    max_iter=1000,
    random_state=42
)

model.fit(X_train_processed, y_train)

print("Model training completed!")
```

```python
y_prob = model.predict_proba(X_test_processed)[:, 1]

print("First 10 predicted probabilities:")
print(y_prob[:10])
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", roc_auc)
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.figure(figsize=(8, 6))

plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Hospital Readmission Prediction")
plt.legend()
plt.show()
```

```python
print("Average feature values by readmission status:\n")

print(
    train_df.groupby("readmitted")[
        ["age", "num_procedures", "days_in_hospital", "comorbidity_score"]
    ].mean()
)
```

```python
print("Average feature values by readmission status:\n")

print(
    train_df.groupby("readmitted")[
        ["age", "num_procedures", "days_in_hospital", "comorbidity_score"]
    ].mean()
)
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score

majority_prediction = [0] * len(y_test)

print("Majority Class Accuracy:",
      accuracy_score(y_test, majority_prediction))

print("Logistic Regression ROC-AUC:",
      roc_auc_score(y_test, y_prob))
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score

thresholds_to_test = [0.10, 0.20, 0.30, 0.40, 0.50]

for threshold in thresholds_to_test:

    y_pred = (y_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()

    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)

    print(f"\nThreshold: {threshold}")
    print("True Negatives :", tn)
    print("False Positives:", fp)
    print("False Negatives:", fn)
    print("True Positives  :", tp)
    print("Precision       :", round(precision, 3))
    print("Recall          :", round(recall, 3))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

threshold = 0.20

y_pred = (y_prob >= threshold).astype(int)

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
print("Minimum probability:", y_prob.min())
print("Maximum probability:", y_prob.max())
print("Mean probability:", y_prob.mean())
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

threshold = 0.20
y_pred = (y_prob >= threshold).astype(int)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Not Readmitted", "Readmitted"]
)

plt.title("Confusion Matrix - Hospital Readmission")
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_prob)

print("Hospital Readmission Model Results")
print("-----------------------------------")
print("ROC-AUC :", round(roc_auc, 4))
print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))
print("Threshold:", threshold)
```

```python
test_processed = preprocessor.transform(test_df)

test_prob = model.predict_proba(test_processed)[:, 1]

test_pred = (test_prob >= threshold).astype(int)

print("Test predictions:", len(test_pred))
print("Predicted 0:", (test_pred == 0).sum())
print("Predicted 1:", (test_pred == 1).sum())
```

```python
submission = sample_submission.copy()

submission["readmitted"] = test_pred

submission.to_csv("submission.csv", index=False)

print("Submission file created!")
display(submission.head())
```
